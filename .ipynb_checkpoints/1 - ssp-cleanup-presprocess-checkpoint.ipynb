```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

# -----------------------------
# 1. Load Data
# -----------------------------
df = pd.read_csv("/home/rizakhan/code/Dr-Shakunt-Jadav/ssp/raw_data/startup_data.csv")

# -----------------------------
# 2. Handle Missing Values
# -----------------------------
missing_counts = df.isna().sum().sort_values(ascending=False)

# Drop columns with >50% missing values
threshold = 0.5
cols_to_drop = missing_counts[missing_counts > len(df) * threshold].index
df = df.drop(columns=cols_to_drop)

# Impute numeric missing values with median
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Impute categorical missing values with "unknown"
categorical_cols = df.select_dtypes(include=["object", "category"]).columns
for col in categorical_cols:
    df[col] = df[col].fillna("unknown")

# -----------------------------
# 3. Fix Inconsistent Formats
# -----------------------------
# Normalize categorical text
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()

# Convert date columns (update list if needed)
date_cols = ["founded_at"]
for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

# -----------------------------
# 4. Remove Duplicates
# -----------------------------
df = df.drop_duplicates()

# -----------------------------
# 5. Normalize Categorical Labels (example mapping)
# -----------------------------
normalization_map = {
    "tech": "technology",
    "technology": "technology",
    "fintech": "financial technology",
    "fin tech": "financial technology"
}

for col in categorical_cols:
    df[col] = df[col].replace(normalization_map)

# -----------------------------
# 6. Detect Outliers (Z-score)
# -----------------------------
z_scores = df[numeric_cols].apply(zscore)
outlier_counts = (np.abs(z_scores) > 3).sum()

print("Outliers detected per column:")
print(outlier_counts)

# Optional: cap outliers (winsorization)
for col in numeric_cols:
    lower = df[col].quantile(0.01)
    upper = df[col].quantile(0.99)
    df[col] = np.clip(df[col], lower, upper)

# -----------------------------
# 7. Save Cleaned Dataset
# -----------------------------
output_path = "/home/rizakhan/code/Dr-Shakunt-Jadav/ssp/raw_data/startup_data_clean.csv"
df.to_csv(output_path, index=False)

print(f"Data cleaning complete. Cleaned file saved to:\n{output_path}")
```
