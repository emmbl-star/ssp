```python
# Import required libraries

import pandas as pd
import numpy as np
import os

# Visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning libraries
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
# Load dataset
df = pd.read_csv("/home/shakunt/code/Dr-Shakunt-Jadav/ssp/raw_data/startup data.csv")

df.head()
```

```python
# Display dataset dimensions

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.tail()
```

```python
# Display column information

df.info()
```

```python
# Display missing values

df.isnull().sum().sort_values(ascending=False)
```

```python
# Target variable

y = df['status']

# Check target distribution

y.value_counts()
```

```python
# Remove target and leakage variables

X = df.drop(columns=[
    'status',
    'labels'
])
```

```python
# Remove text and identifier columns

drop_cols = [
    'state_code',
    'zip_code',
    'id',
    'city',
    'Unnamed: 6',
    'name',
    'founded_at',
    'closed_at',
    'first_funding_at',
    'last_funding_at',
    'state_code.1',
    'category_code',
    'object_id'
]

X = X.drop(columns=drop_cols)
```

```python
# Replace missing values with 0

X = X.fillna(0)

# Verify no missing values remain

X.isnull().sum().sum()
```

```python
# Split data into training and testing datasets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Create baseline Random Forest model

rf_model = RandomForestClassifier(
    random_state=42
)

# Train model

rf_model.fit(X_train, y_train)
```

```python
# Generate predictions

y_pred = rf_model.predict(X_test)
```

```python
# Calculate accuracy

accuracy = accuracy_score(y_test, y_pred)

print(f"Baseline Accuracy: {accuracy:.2%}")
```

```python
# Display classification report

print(classification_report(y_test, y_pred))
```

```python
# Plot confusion matrix

ConfusionMatrixDisplay.from_estimator(
    rf_model,
    X_test,
    y_test
)

plt.show()
```

```python
# Create feature importance dataframe

feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_model.feature_importances_
})

# Display top features

feature_importance.sort_values(
    by='Importance',
    ascending=False
).head(10)
```

```python
# Import Dummy Classifier
# This model ignores all features and makes predictions using a simple strategy.
# It serves as a baseline that our real models should outperform.

from sklearn.dummy import DummyClassifier
```

```python
# Create a dummy classifier
# Strategy = 'most_frequent' means the model always predicts
# the most common class in the training dataset.

dummy_model = DummyClassifier(
    strategy='most_frequent'
)
```

```python
# Train the dummy model using the training data

dummy_model.fit(X_train, y_train)
```

```python
# Generate predictions on the test dataset

dummy_pred = dummy_model.predict(X_test)
```

# Dummy Model

The Dummy Classifier provides a benchmark for model performance.

It does not learn patterns from the data. Instead, it uses a simple strategy
such as always predicting the most common class.

Any machine learning model developed in this project should outperform the
Dummy Classifier to demonstrate predictive value.

```python
# Calculate the accuracy of the dummy model

dummy_accuracy = accuracy_score(
    y_test,
    dummy_pred
)

print(f"Dummy Model Accuracy: {dummy_accuracy:.2%}")
```

# Dummy Model Results

A Dummy Classifier was trained as a baseline benchmark.

The model does not learn from any startup characteristics and simply predicts
the most frequent class in the dataset.

The Dummy Model achieved an accuracy of 60.54%.

Any machine learning model developed in this project should outperform this
baseline to demonstrate predictive capability.
